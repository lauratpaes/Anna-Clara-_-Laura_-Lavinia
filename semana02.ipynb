```python
# 1. Organização inicial, carregar os dados e conferir a estrutura
import pandas as pd
import matplotlib.pyplot as plt

# 1.1 Carregar a planilha do Excel
df = pd.read_excel('dados_semana02.xlsx')

# 1.2 Exibir as primeiras 5 linhas
df.head()
```

```python
# 2. Gráfico de pizza
import matplotlib.pyplot as plt

# 2.1 Limpa espaços extras nos nomes das colunas
df.columns = df.columns.str.strip()

# 2.2 Conta quantas vezes cada tratamento aparece
contagem = df['Tratamento'].value_counts()

# 2.3 Lista com uma cor para cada fatia
minhas_cores = ['#ceffa2', '#fefe7c', '#fddee2']

# 2.4 Desenha o gráfico de pizza
plt.figure(figsize=(6, 6))
plt.pie(contagem, labels=contagem.index, autopct='%1.1f%%', startangle=90, colors=minhas_cores)
plt.title('Proporção de Amostras por Tratamento', fontsize=12, fontweight='bold')
plt.axis('equal')

plt.show()
```

```python
# 3.Gráfico de barras
# 3.1 Comprimento crescimento radicular
import seaborn as sns
import matplotlib.pyplot as plt

# 3.1.1 Ajusta o tamanho do gráfico
plt.figure(figsize=(10, 5))

# 3.1.2 Desenha o gráfico de barras
sns.barplot(data=df, x='Tratamento', y='Comprimento crescimento radicular', errorbar=None, color='#e29aff')

# 3.1.3 Adiciona títulos e rótulos dos eixos
plt.title('Comprimento crescimento radicular por Tratamento', fontsize=12, fontweight='bold')
plt.xlabel('Tipo de Tratamento', fontsize=10)
plt.ylabel('Comprimento crescimento radicular (Cm)', fontsize=10)
plt.xticks(rotation=0, fontsize=7)

plt.tight_layout()
plt.show()
```

```python
# 3.Gráfico de barras
# 3.2 Comprimento crescimento caulículo
import seaborn as sns
import matplotlib.pyplot as plt

# 3.2.1 Ajusta o tamanho do gráfico
plt.figure(figsize=(10, 5))

# 3.2.2 Desenha o gráfico de barras
sns.barplot(data=df, x='Tratamento', y='Comprimento crescimento caulículo', errorbar=None, color='#a6c7ff')

# 3.2.3 Adiciona títulos e rótulos dos eixos
plt.title('Comprimento crescimento caulículo por Tratamento', fontsize=12, fontweight='bold')
plt.xlabel('Tipo de Tratamento', fontsize=10)
plt.ylabel('Comprimento crescimento caulículo (Cm)', fontsize=10)
plt.xticks(rotation=0, fontsize=7)

plt.tight_layout()
plt.show()
```

```python
# 4. Boxplot
# 4.1 Analise do crescimento radicular por tratamento
import seaborn as sns
import matplotlib.pyplot as plt

# 4.1.1 Ajusta o tamanho do gráfico
plt.figure(figsize=(10, 5))

# 4.1.2 Desenha o boxplot para comparar o crescimento radicular por tratamento
sns.boxplot(data=df, x='Tratamento', y='Comprimento crescimento radicular', color='#e29aff')

# 4.1.3 Adiciona títulos e rótulos dos eixos
plt.title('Distribuição do Crescimento da Radícula por Tratamento', fontsize=12, fontweight='bold')
plt.xlabel('Tipo de Tratamento', fontsize=10)
plt.ylabel('Comprimento da radicula (cm)', fontsize=10)
plt.xticks(rotation=0, fontsize=8)

plt.tight_layout()
plt.show()
```

```python
# 4. Boxplot
# 4.2 Analise do crescimento caulículo por tratamento
import seaborn as sns
import matplotlib.pyplot as plt

# 4.2.1 Ajusta o tamanho do gráfico
plt.figure(figsize=(10, 5))

# 4.2.2 Desenha o boxplot para comparar o crescimento caulíco por tratamento
sns.boxplot(data=df, x='Tratamento', y='Comprimento crescimento caulículo', color='#a6c7ff')

# 4.2.3 Adiciona títulos e rótulos dos eixos
plt.title('Distribuição do Crescimento do Caulículo por Tratamento', fontsize=12, fontweight='bold')
plt.xlabel('Tipo de Tratamento', fontsize=10)
plt.ylabel('Comprimento do Caulículo (cm)', fontsize=10)
plt.xticks(rotation=0, fontsize=8)

plt.tight_layout()
plt.show()
```

```python
# 5. Histograma
# 5.1 Analise da distribuição de comprimento raducular por tratamento
import seaborn as sns
import matplotlib.pyplot as plt

# 5.1.1 Ajusta o tamanho do gráfico
plt.figure(figsize=(10, 5))

# 5.1.2 Desenha o histograma do crescimento da radícula
sns.histplot(df['Comprimento crescimento radicular'], kde=True, color='#e29aff')

# 5.1.3 Adiciona títulos e rótulos dos eixos
plt.title('Distribuição do Comprimento Radicular das Plântulas', fontsize=12, fontweight='bold')
plt.xlabel('Comprimento Radicular (cm)', fontsize=10)
plt.ylabel('Frequência (Quantidade de Observações)', fontsize=10)

plt.tight_layout()
plt.show()
```
