#Conjunto de dados biológicos

```python
import pandas as pd
df = pd.DataFrame({
    "animal": [
        'plebeia sp.', 'plebeia sp.', 'formigas', 'abelha-jataí',
        'T. spinipes', 'D. plexippus', 'formigas',
        'abelha-jataí', 'beija-flor-tesoura', 'D. plexippus',
        'formigas', 'D. plexippus', 'E. macroura',
        'H. hermes', 'abelha-jataí', 'Tribo Augochlorini',
        'A. mellifera', 'abelha-jataí', 'T. spinipes', 'abelha-jataí'
    ],  # nominal
    "número_visitas": [
       1, 2, 2, 15, 1, 1, 4, 3, 1, 2, 4, 1, 2, 3, 15, 4, 2, 17, 11, 21
    ],  # discreta
    "temperatura": [
        21, 22, 27, 25, 25, 25, 23, 23, 23, 22, 22, 23, 22, 23, 22,
        22, 22, 23, 23, 25
    ]  # contínua
})
df
```

#Tabelas de frequências

```python
# 1. Frequência simples da variável nominal: animal
freq_animal = df['animal'].value_counts()
freq_animal
```

A partir desta tabela pode-se inferir que "abelha-jataí" foi o animal que mais se repetiu ao longo das observações, já que possui um total de 5 repetições. Em comparação, outras espécies de abelhas tiveram baixa frequência, como "A. mellifera" e "Tribo Auchlorini".

```python
# 1.1 Frequência simples da variável discreta: número de visitas
freq_número_visitas = df['número_visitas'].value_counts()
freq_número_visitas
```

Esta tabela nos revela que a ocorrência de espécies em menor quantidade é mais frenquente, já que observações com apenas 1 e 2 visitantes apresentaram a maior repetição, de valor igual a 5. Em contrapartida, observações com maior número de visitantes florais tiveram menor frequência, com ocorreu com a repetição única de 21, 17 e 11 visitantes.

```python
# 2.1 frequência relativa da variável nominal: animal
freq_animal_rel = df['animal'].value_counts(normalize=True) * 100
freq_animal_rel
```

Assim como na tabela de frequência simples, é possível inferir que a maior proporção de visitantes florais é pertencente ao animal "abelha jataí", com 25%, seguido de "formigas" e "D. plexippus", com 15% cada. Além disso, nota-se que metade das classes apresentaram a mesma proporção de 5%, o que mostra que o predomínio de certas espécies no ambiente em relação a outras.

```python
# 2.2 Frequência da variável discreta: número de visitas
freq_número_visitas_rel = df['número_visitas'].value_counts(normalize=True) * 100
freq_número_visitas_rel
#
```

A tabela demonstra, assim como na frequência simples, o domínio do número de visitas em '1" e "2', que somados resultam em 50% dos casos. Em contrapartida, ocasiões em que um maio número de visitantes foram observados possuem menor proporção, como em "17", "11" e "21", todos com 5%.

# Gráficos

##Gráfico de barras

```python
import matplotlib.pyplot as plt
import seaborn as sns

plt.figure(figsize=(15, 4))
sns.countplot(data=df, x="animal", color="blue")
plt.title("Frequência de espécies")
plt.xlabel("Animal")
plt.ylabel("Frequência")
plt.show()
```

#Histograma

```python
plt.figure(figsize=(6, 4))
sns.histplot(
    data=df,
    x="temperatura",
    bins=[15, 20, 25, 30, 35],
    kde=False
)
plt.title("Distribuição de visitantes florais por faixa de temperatura")
plt.xlabel("Temperatura (°C)")
plt.ylabel("Frequência")
plt.show()
```

#Gráfico de dispersão

```python
plt.figure(figsize=(9, 6))
sns.scatterplot(
    data=df,
    x="temperatura",
    y="número_visitas",
    hue="animal"
)
plt.title("Número de visitantes vs. temperatura")
plt.xlabel("temperatura (°C)")
plt.ylabel("Número de visitantes")
plt.yticks(range(0, df['número_visitas'].max() + 3, 2)) # Ajusta os ticks do eixo y para ir de 2 em 2
plt.show()
```

#Relatório

-Gráfico de barras: Este gráfico demonstra o predomínio de abelhas jataí nas observações, com uma frequeência de 5, ou seja, cinco repetições ao longo dos dados, seguido por formigas e D. plexippus. Em geral, este gráfico constitui uma visualização da tabela de frequência simples da variável nominal "animal", sendo possível inferir as mesmas informações que esta.

-Histograma: Este gráfico está centrado na nossa variável contínua "temperatura", demonstrando a frequência de espécie correlacionada com a faixa de temperatura. Nele é possível inferir que há uma maior frequência de visitantes florais entre 20°C e 25°C, além de uma menore frequência entre 25°C e 27°C. Não foram observados visitantes em temperaturas acima de 30°C e abaixo de 20°C.

-Gráfico de dispersão: Este gráfico correlaciona as variáveis "temperatura" e "número de visitantes". Nota-se que há uma maior quantidade de ocorrências na temperatura de 23°C, além de que há um acúmulo de número de visitantes abaixo do ponto 5 do eixo y. Ademais, infere-se que o animal predominante "abelha-jataí" foi observado entre 22°C e 25°C, com número de visitantes sempre igual ou superior a 15. É notável o domínio de espsécies de abelhas nas observações com "Abelha-jatai" e "T. spinipes", esta com uma única ocorrência no valor acima de 10 visitantes.
