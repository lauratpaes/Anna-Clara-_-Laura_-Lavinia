## Sobre mediana e quartis

A mediana é um importante conceito aplicado na atividade, já que o cálculo de quartis tem como base a mediana. Este conceito refer-se ao valor do meio de uma série de valores organizados em ordem crescente ou decrescente. Por se tratar de um ról, e não de uma distribuição de frequência, a mediana é o valor exato centralizado no meio do ról da variável diâmetro a altura do peito. Por possuir 3891 valores, a mediana será exatamente o valor na posição 1946º, de valor 124cm, ficando 1945 valores antes, ou seja, menores ou iguais a 124cm, e 1945 depois, maiores ou iguais a 124cm.

A mediana, por ser uma média de posição, diz bastante sobre o que acontece na zona do meio da distribuição, mas não informa muito sobre o que se passa antes ou depois. Para isso, existe uma definição derivada do conceito de mediana: quartis. Este conceito servem para a caracterização de um determinada distribuição, além de comparar duas ou mais distribuições da mesma variável. São definidos 3 quartis, em que inicialmente se define o segundo quartil, o qual consiste no valor da mediana. O segundo quartil define o ponto central da distribuição, em que 50% dos valores estarão abaixo deste e 50% estarão acima. A partir daí é calculado o primeiro quartil, que consiste em um ponto que demarca que 25% dos valores estarão abaixo deste ponto, ou seja, é uma mediana dos primeiros 50% dos valores, segregando em os 25% menores e 75% maiores. Em seguida, é calculado o terceiro quartil, que segue a mesma linha de raciocínio: é determinada a mediana dos 50% finais, segregando os 75% valores menores que este ponto e os 25% maiores que ele.

Assim, utilizando nossos dados, o primeiro quartil possui valor de 103cm, o segundo de 124cm e o terceiro de 155cm.

```python
import matplotlib.pyplot as plt
import seaborn as sns

plt.figure(figsize=(8, 6))
sns.boxplot(x=df['DAP'], orient='h')
plt.title('Box Plot of DAP Variable')
plt.xlabel('DAP')
plt.grid(True, linestyle='--', alpha=0.7)
plt.show()
```

O bloxpot é um gráfico gerado utilizando os cálculos de quartis, que permite observar a distribuição de uma variável Na imagem, sua centralidade, dispersão e valores atípicos. Observa-se que o gráfico consiste em uma caixa colorida que demarca o primeiro quartil à esquerda, o terceiro quartil à direita e o segundo quartil dentro da caixa. A largura desta caixa colorida é definida pela subtração entre o terceiro quartil e o primeiro quartil, sendo esta a amplitude interquartílica. No caso dos nossos dados, ela possui valor igual a 52cm, e mostra a faixa em que 50% dos dados estão localizados.

No entanto, esta colocação de normalidade estreteita bastante a tolerância e exclui 50% dos valores que seriam definidos como atípicos. Para aumentar esta tolerância, são definidos os "bigodes", que se estendem até um limite para fora da caixa, calculados com base no primeiro quartil, terceiro quartil e na amplitude interquartílica. Estes bigodes mostram a variabilidade geral dos dados e defidem quais valores são considerados típicos.

Desta forma, os valores que se encontram fora da caixa azul e dos bigodes são considerados atípicos, ou seja, que são valores significativamente diferentes da maioria dos dados (que se encontram dentro do limite). Estes valores atípicos são representados pelas bolinhas no nosso gráfico, que apenas aparecem fora dos bigodes, tanto antes quanto depois dos bigodes.

## Como determinamos se há observações atípicas em uma base de dados?

Aplicando tudo falado até agora, concluímos que para esta determinação é preciso calcular os quartis e aplicá-los em um bloxpot, a fim de definir uma margem de tolerância de 1,5 vezes para menos e para mais da amplitude interquartílica com base nos valores dos peimeiro e terceiro quartis. Assim, o que fica exterior a esta margem são chamados observações atípicas, dados que se diferem em grande margem da maioria.
