## Exercício 1.

A massa corporal de uma espécie de pequeno mamífero apresenta distribuição normal:

• μ=420 g

• σ=55 g

Determine:
1. A probabilidade de um indivíduo pesar menos de 350 g.

R: $P(X < 350 g)$ = 10.16%

2. A probabilidade de pesar mais de 500 g.

R: $P(X > 500 g)$ = 7.29%

3. A probabilidade de pesar entre 400 e 480 g.

R: $P(400 g < X < 480 g)$ = 50.43%

4. Represente graficamente as três probabilidades.

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import norm

mu = 420
sigma = 55

# 1. Probabilidade de pesar menos de 350 g
p_menos_350 = norm.cdf(350, loc=mu, scale=sigma)

# 2. Probabilidade de pesar mais de 500 g
p_mais_500 = 1 - norm.cdf(500, loc=mu, scale=sigma)

# 3. Probabilidade de pesar entre 400 e 480 g
p_entre_400_480 = norm.cdf(480, loc=mu, scale=sigma) - norm.cdf(400, loc=mu, scale=sigma)

print(f"1. P(X < 350 g) = {p_menos_350:.4f} ({p_menos_350 * 100:.2f}%)")
print(f"2. P(X > 500 g) = {p_mais_500:.4f} ({p_mais_500 * 100:.2f}%)")
print(f"3. P(400 g < X < 480 g) = {p_entre_400_480:.4f} ({p_entre_400_480 * 100:.2f}%)")
```

```python
# 4. Representação gráfica das três probabilidades
x = np.linspace(mu - 4*sigma, mu + 4*sigma, 1000)
y = norm.pdf(x, loc=mu, scale=sigma)

fig, axes = plt.subplots(1, 3, figsize=(18, 5), sharey=True)

# Gráfico 1: P(X < 350)
axes[0].plot(x, y, 'r-', lw=2, label='Normal PDF')
x_fill1 = np.linspace(mu - 4*sigma, 350, 500)
y_fill1 = norm.pdf(x_fill1, loc=mu, scale=sigma)
axes[0].fill_between(x_fill1, y_fill1, color='red', alpha=0.3, label=f'P(X < 350)\n= {p_menos_350:.4f}')
axes[0].set_title('P(X < 350 g)')
axes[0].set_xlabel('Peso (g)')
axes[0].set_ylabel('Densidade de Probabilidade')
axes[0].legend()
axes[0].grid(True, linestyle='--', alpha=0.5)

# Gráfico 2: P(X > 500)
axes[1].plot(x, y, 'g-', lw=2, label='Normal PDF')
x_fill2 = np.linspace(500, mu + 4*sigma, 500)
y_fill2 = norm.pdf(x_fill2, loc=mu, scale=sigma)
axes[1].fill_between(x_fill2, y_fill2, color='green', alpha=0.3, label=f'P(X > 500)\n= {p_mais_500:.4f}')
axes[1].set_title('P(X > 500 g)')
axes[1].set_xlabel('Peso (g)')
axes[1].legend()
axes[1].grid(True, linestyle='--', alpha=0.5)

# Gráfico 3: P(400 < X < 480)
axes[2].plot(x, y, 'b-', lw=2, label='Normal PDF')
x_fill3 = np.linspace(400, 480, 500)
y_fill3 = norm.pdf(x_fill3, loc=mu, scale=sigma)
axes[2].fill_between(x_fill3, y_fill3, color='blue', alpha=0.3, label=f'P(400 < X < 480)\n= {p_entre_400_480:.4f}')
axes[2].set_title('P(400 g < X < 480 g)')
axes[2].set_xlabel('Peso (g)')
axes[2].legend()
axes[2].grid(True, linestyle='--', alpha=0.5)

plt.suptitle('Distribuição Normal da Massa Corporal (μ=420g, σ=55g)', fontsize=16, y=1.02)
plt.tight_layout()
plt.show()
```

## Exercício 2.

O comprimento corporal desses mamíferos segue uma distribuição normal com:

• μ=150 cm

• σ=10,5 cm

Calcule:
1. O percentil 80.

R: $P80$ = 158.84 cm

2. O valor correspondente a 10% dos indivíduos apresentarem o comprimento superior a esse
valor.

R: $P90$ = 163.46 cm

3. Faça um único gráfico indicando os três percentis.

Interprete biologicamente cada resultado. Por exemplo: qual comprimento separa os 10% maiores
indivíduos dos demais?

R: O percentil 80 significa que 80% dos dados possuem valor menor ou igual à 158.84 cm, ou seja, 80% dos pequenos mamíferos possuem comprimento corporal até 158.84 cm, enquanto os 20% restantes estão acima deste valor. Em contrapartida, o o valor que representa os 10% maiores que o percentil 80 é de 163.46 cm, que também representa o percentil 90, ou seja, 90% dos mamíferos possuem comprimento corporal igual ou menor a este valor. Ao mesmo tempo, este valor separa os 10% com maior comprimento corporal, separação biológica importante que demarca a extremidade da distribuição para a variável,

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import norm

mu_comp = 150
sigma_comp = 10.5

# 1. Percentil 80 (P80)
p80 = norm.ppf(0.80, loc=mu_comp, scale=sigma_comp)

# 2. Valor acima do qual estão apenas 10% dos indivíduos (Percentil 90 ou P90)
p90 = norm.ppf(0.90, loc=mu_comp, scale=sigma_comp)

print(f"1. O percentil 80 (P80) é: {p80:.2f} cm")
print(f"2. O valor que deixa os 10% maiores acima dele (P90) é: {p90:.2f} cm")
```

```python
# 3. Gráfico único indicando os percentis calculados (incluindo a mediana/P50 para referência)
p50 = norm.ppf(0.50, loc=mu_comp, scale=sigma_comp)

x_comp = np.linspace(mu_comp - 4*sigma_comp, mu_comp + 4*sigma_comp, 1000)
y_comp = norm.pdf(x_comp, loc=mu_comp, scale=sigma_comp)

plt.figure(figsize=(12, 6))
plt.plot(x_comp, y_comp, 'k-', lw=2, label='Distribuição de Comprimento')

# Linhas verticais para os percentis
plt.axvline(p80, color='pink', linestyle='--', lw=2, label=f'P80 (Percentil 80) = {p80:.2f} cm')
plt.axvline(p90, color='red', linestyle='--', lw=2, label=f'P90 (Superior a 10%) = {p90:.2f} cm')
plt.axvline(p50, color='blue', linestyle=':', lw=1.5, label=f'P50 (Mediana/Média) = {p50:.2f} cm')

# Preenchimento visual das áreas
plt.fill_between(x_comp, y_comp, where=(x_comp >= p90), color='red', alpha=0.3, label='10% Maiores Comprimentos')
plt.fill_between(x_comp, y_comp, where=(x_comp >= p80) & (x_comp < p90), color='orange', alpha=0.2, label='Entre P80 e P90')

plt.title('Distribuição Normal do Comprimento Corporal (μ=150cm, σ=10.5cm)', fontsize=14)
plt.xlabel('Comprimento (cm)', fontsize=12)
plt.ylabel('Densidade de Probabilidade', fontsize=12)
plt.legend(loc='upper left')
plt.grid(True, linestyle='--', alpha=0.5)
plt.tight_layout()
plt.show()
```

## Exercício 3.

Uma população de aves apresenta massa corporal normalmente distribuída com:

• μ=72 g

• σ=8 g

Em Python:
1. Calcule P(X≤60).

R: $P(X <= 60 g)$ = 6.68%

2. Calcule P(X≥85).

R:  $P(X >= 85 g)$ = 5.21%

3. Calcule P(65≤X≤80).

R: $P(65 g <= X <= 80 g)$ = 65.06%

4. Calcule os percentis 90, 95 e 99.

R: P90 = 82.25 g

   P95 = 85.16 g

   P99 = 90.61 g

5. Produza um gráfico da distribuição normal mostrando as três regiões de probabilidade e os
três percentis.

6. Escreva uma interpretação ecológica curta dos resultados.

R: A probabilidade de encontrar uma ave com massa corporal menor ou igual a 60g é de 6.68%. Biologicamente, isso significa que uma parcela baixa da população são consideradas leves e pode implicar que são indivíduos juvenis. Já a probabilidade de encontrar uma ave com peso corporal maior ou igual a 85g é de 5.21%. Isso pode significar que a espécie em questão não chega a valores de massa altos ou ainda pode ser uma fêmea em período de postura de ovos. Já a probabilidade de encontrar uma ave dentro do intervalo de massa corporal de 65g a 80g é de 65.06%. Isso significa que, na distribuição Normal, a maioria dos indivíduos estão concentrados nesta faixa central, o que implica biologicamente que este intervalo é a massa corporal mais recorrente da espécie. Já os percentis estabelecem marcas importantes: P90 = 82.25g significa que 90% da população possui massa igual ou menor a este valor, enquanto apenas 10% está acima dele; P95 = 85.16 g significa que 5% da população possui massa corporal superior a este valor; por fim, P99 = 90.61 g diz que apenas 1% da população possui massa maior que 90.61g, sendo valores outliners.

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import norm

# Parâmetros da população de aves
mu_aves = 72
sigma_aves = 8

# 1. P(X <= 60)
p_60 = norm.cdf(60, loc=mu_aves, scale=sigma_aves)

# 2. P(X >= 85)
p_85 = 1 - norm.cdf(85, loc=mu_aves, scale=sigma_aves)

# 3. P(65 <= X <= 80)
p_65_80 = norm.cdf(80, loc=mu_aves, scale=sigma_aves) - norm.cdf(65, loc=mu_aves, scale=sigma_aves)

# 4. Percentis 90, 95 e 99
p90_aves = norm.ppf(0.90, loc=mu_aves, scale=sigma_aves)
p95_aves = norm.ppf(0.95, loc=mu_aves, scale=sigma_aves)
p99_aves = norm.ppf(0.99, loc=mu_aves, scale=sigma_aves)

print(f"1. P(X <= 60 g) = {p_60:.4f} ({p_60 * 100:.2f}%)")
print(f"2. P(X >= 85 g) = {p_85:.4f} ({p_85 * 100:.2f}%)")
print(f"3. P(65 g <= X <= 80 g) = {p_65_80:.4f} ({p_65_80 * 100:.2f}%)")
print(f"4. Percentil 90 = {p90_aves:.2f} g")
print(f"   Percentil 95 = {p95_aves:.2f} g")
print(f"   Percentil 99 = {p99_aves:.2f} g")
```

```python
# 5. Produzir um gráfico da distribuição normal com as três regiões e os três percentis
x_aves = np.linspace(mu_aves - 4*sigma_aves, mu_aves + 4*sigma_aves, 1000)
y_aves = norm.pdf(x_aves, loc=mu_aves, scale=sigma_aves)

plt.figure(figsize=(14, 7))
plt.plot(x_aves, y_aves, 'k-', lw=2.5, label='Distribuição de Massa Corporal (Aves)')

# Colorir as três regiões de probabilidade
plt.fill_between(x_aves, y_aves, where=(x_aves <= 60), color='red', alpha=0.3, label=f'P(X <= 60 g) = {p_60*100:.1f}%')
plt.fill_between(x_aves, y_aves, where=(x_aves >= 85), color='blue', alpha=0.3, label=f'P(X >= 85 g) = {p_85*100:.1f}%')
plt.fill_between(x_aves, y_aves, where=(x_aves >= 65) & (x_aves <= 80), color='green', alpha=0.3, label=f'P(65 <= X <= 80 g) = {p_65_80*100:.1f}%')

# Marcar os três percentis
plt.axvline(p90_aves, color='purple', linestyle='--', lw=2, label=f'P90 = {p90_aves:.2f} g')
plt.axvline(p95_aves, color='darkorange', linestyle='-.', lw=2, label=f'P95 = {p95_aves:.2f} g')
plt.axvline(p99_aves, color='crimson', linestyle=':', lw=2, label=f'P99 = {p99_aves:.2f} g')

plt.title('Distribuição da Massa Corporal de Aves (μ=72g, σ=8g)', fontsize=14)
plt.xlabel('Massa Corporal (g)', fontsize=12)
plt.ylabel('Densidade de Probabilidade', fontsize=12)
plt.legend(loc='upper left', fontsize=10)
plt.grid(True, linestyle='--', alpha=0.5)
plt.tight_layout()
plt.show()
```
